```python
import sys
sys.path.append("./functions/")

from functions.helpers import load_csv_data
from functions.implementations import lr_inference
from functions.helpers import create_csv_submission
import numpy as np

# Load Model
model_name = "./models/simple_model_try8.txt"
sub_name = "submission_model_8.csv"
weights = np.loadtxt(model_name, delimiter=",")

# Load Data
x_test = np.loadtxt("./processed_dataset/X_test_processed.csv", delimiter=",")
bias = np.ones(len(x_test)).reshape(len(x_test), 1)
x_test = np.concatenate((x_test, bias), axis=1) # Add the bias term

weights_feature = np.loadtxt("./models/simple_model_try7.txt", delimiter=",")

res = []
for i in range(len(x_test)):
    classification = lr_inference(x_test[i:(i+1)], weights_feature)
    classification = classification[0]
    res.append(classification)

x_test = np.concatenate((x_test, np.array(res).reshape(len(res), 1)), axis=1)
```

```python
res = []
for i in range(len(x_test)):
    classification = lr_inference(x_test[i:(i+1)], weights)
    classification = classification[0]
    if classification == 0:
        classification = -1
    else:
        classification = 1
    res.append(classification)
```

```python
# Save the result
ids = [x for x in range(328135, 328135+len(res))]
create_csv_submission(ids, res, sub_name)
```
